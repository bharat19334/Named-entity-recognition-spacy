In this project, we are using an NER model to read text, find important entities, and classify them into categories like PERSON, ORG, GPE, DATE, and MONEY.

```python
%pip install https://github.com/explosion/spacy-models/releases/download/en_core_web_sm-3.8.0/en_core_web_sm-3.8.0-py3-none-any.whl
```

```python
import spacy
from spacy import displacy
```

```python
NER = spacy.load("en_core_web_sm")
```

```python
text = NER("Elon Musk works at SpaceX in USA.")

for word in text.ents:
    print(word.text, word.label_)
```

```python
raw_text = (
    "The Indian Space Research Organisation is the national space "
    "agency of India, headquartered in Bengaluru. It operates under "
    "Department of Space which is directly overseen by the Prime "
    "Minister of India while Chairman of ISRO acts as executive of DOS."
)
text1 = NER(raw_text)
```

```python
print("\nEntity Label Explanation:")

print("ORG:", spacy.explain("ORG"))
print("GPE:", spacy.explain("GPE"))

displacy.render(text1, style="ent", jupyter=True)
```

```python
raw_text2 = (
    "The Mars Orbiter Mission (MOM), informally known as Mangalyaan, "
    "was launched into Earth orbit on 5 November 2013 by the Indian "
    "Space Research Organisation (ISRO) and has entered Mars orbit "
    "on 24 September 2014. India thus became the first country to "
    "enter Mars orbit on its first attempt. It was completed at a "
    "record low cost of $74 million."
)
text2 = NER(raw_text2)
```

```python
print("\nSecond Text - Named Entities:\n")

for word in text2.ents:
    print(word.text, word.label_)
```

```python
print("\nEntity Label Explanation:\n")

print("PERSON:", spacy.explain("PERSON"))
print("ORG:", spacy.explain("ORG"))
print("GPE:", spacy.explain("GPE"))
print("PRODUCT:", spacy.explain("PRODUCT"))
print("LOC:", spacy.explain("LOC"))
print("DATE:", spacy.explain("DATE"))
print("ORDINAL:", spacy.explain("ORDINAL"))
print("MONEY:", spacy.explain("MONEY"))
```

```python
print("\nSecond NER Visualization:")

displacy.render(text2, style="ent", jupyter=True)
```

### News Article NER

```python
from bs4 import BeautifulSoup
import requests
import re
```

```python
URL = (
    "https://www.zeebiz.com/markets/currency/"
    "news-cryptocurrency-news-today-june-12-bitcoin-dogecoin-"
    "shiba-inu-and-other-top-coins-prices-and-all-latest-"
    "updates-158490"
)
```

```python
html_content = requests.get(URL).text
```

```python
%pip install beautifulsoup4 requests lxml
```

```python
soup = BeautifulSoup(html_content, "lxml")
```

```python
body = soup.body.text
```

```python
body = body.replace("\n", " ")
body = body.replace("\t", " ")
body = body.replace("\r", " ")
body = body.replace("\xa0", " ")

body = re.sub(r"[^\w\s]", "", body)
```

```python
print(body[1000:1500])
```

```python
text3 = NER(body)
```

```python
print("\nNews Article - Named Entities:\n")

for word in text3.ents:
    print(word.text, word.label_)
```

```python
displacy.render(text3, style="ent", jupyter=True)
```
